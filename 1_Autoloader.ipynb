#Increamental Data Loading Using Autoloader

```python
%sql
CREATE SCHEMA netflix_catalog.net_schema;
 
```

```python
checkpoint_location = "abfss://silver@netflixsharsh1.dfs.core.windows.net/checkpoint"
```

```python
df=spark.readStream\
    .format("cloudFiles")\
    .option("cloudFiles.format","csv")\
    .option("cloudFiles.schemaLocation",checkpoint_location)\
    .load("abfss://raw@netflixsharsh1.dfs.core.windows.net")
```

```python
df.writeStream\
 .option("checkpointLocation", checkpoint_location)\
 .trigger(processingTime='10 seconds')\
 .start("abfss://bronze@netflixsharsh1.dfs.core.windows.net/netflix_titles")
```
