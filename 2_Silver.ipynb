# Silver Nitebook Lookup Tables


###Parameters

```python
dbutils.widgets.text("sourcefolder","netflix_directors")
dbutils.widgets.text("targetfolder","netflix_directors")
```

###Variables

```python
var_src_folder=dbutils.widgets.get("sourcefolder")
var_tgt_folder=dbutils.widgets.get("targetfolder")
```

```python
print(var_src_folder)
print(var_tgt_folder)
```

```python
df= spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load(f"abfss://bronze@netflixsharsh1.dfs.core.windows.net/{var_src_folder}")
```

```python
df.display()
```

```python
df.write.format("delta")\
    .mode("overwrite")\
    .save(f"abfss://silver@netflixsharsh1.dfs.core.windows.net/{var_tgt_folder}")
```
